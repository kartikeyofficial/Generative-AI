## Standard Library Overview



Python's Standard Library is a vast collection of modules and packages that comes bundled with python, Providing a wide range of functionality out of the box. Here's an overview of some of the most commonly used modules and packages in the python standard Library.

```python
import array
arr = array.array('i',[1,2,3,4])
print(arr)
```

```python
import math
print(math.sqrt(16))
```

```python
# random
import random
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```

```python
## file and directory access
import os
print(os.getcwd())
```

```python
## data Serialization

import json
data = {'name':'kartikey','age':22}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))
```

```python
## csv
import csv

with open('example.csv',mode='w',newline='') as file:
  writer = csv.writer(file)
  writer.writerow(['name','age'])
  writer.writerow(['Kartikey',22])

with open('example.csv', mode='r') as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```
