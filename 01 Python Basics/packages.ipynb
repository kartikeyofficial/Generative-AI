## Importing Module in Python: Modules and Packages

in Python , module and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(25)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,5])


```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
from package.maths import *
print(addition(5,6))
print(substration(6,1))
print(multiplication(2,4))
```
