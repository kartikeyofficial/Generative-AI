## Inheritance in Python


### Single Inheritance

```python
## Inheritance(Single Inheritance)
## Defining the parent class
class Car:
  def __init__(self,windows,doors,wheels,enginetype):
    self.windows = windows
    self.doors = doors
    self.wheels = wheels
    self.enginetype = enginetype
  def drive(self):
    print(f"The person will drive the {self.doors} doors car")
  
car1 = Car(4,4,4)
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self,windows,doors,wheels,enginetype,is_selfdriving):
      super().__init__(windows,doors,wheels,enginetype)
      self.is_selfdriving = is_selfdriving
    
    def selfdriving(self):
      print(f"Tesla Supports self Driving: {self.is_selfdriving}")



```

```python
tesla1 = Tesla(4,4,4,"Electric",True)
tesla1.selfdriving()
```

### Multiple Inheritance

```python
## Multiple Inheritance
## When A Class Inherit From more than one base Class.
## BAse Class 1 
class Animal:
  def __init__(self,name):
    self.name = name
  def speak(self):
    print(f"{self.name} Subscribe must implement this method")

## BAse Class 2
class Pet:
  def __init__(self, owner):
    self.owner = owner

## Derived Class
class Dog(Animal,Pet):
  def __init__(self,name,owner):
    Animal.__init__(self,name)
    Pet.__init__(self,owner)
  def speak(self):
    return f"{self.name} say woof"
  

## Object 
dog = Dog("Husky","Kartikey")
print(dog.speak())
print(f"Owner: {dog.owner}")
```
