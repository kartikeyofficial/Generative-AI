### File Operation- Read And Write Files

```python
## Read a Whole File
with open('example.txt','r') as file:
  content = file.read()
  print(content)
```

```python
## Read A file line by line

with open('example.txt','r') as file:
  for line in file:
    print(line.strip()) ## strip() remove the newline character
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
  file.write('Hello World!\n')
  file.write('this a new line\n')
```

```python
## write a file without(Overwriting)

with open('example.txt','a') as file:
  file.write('\nAppendent Operation taking place')
```

```python
## Writing a list of lines to a file

lines = ['\nfirst line\n','Second line\n','Third Line\n']
with open('example.txt','a') as file:
  file.writelines(lines)
```

### Binary Files

```python
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
  file.write(data)
```

```python
with open('example.bin','rb') as file:
  print(file.read())
```

```python
## writing and then reading the file

with open('example.txt','w+') as file:
  file.write('hey! hello World')
  file.write('\nthis is new Line')

  ## Move the file cursor to the begning
  file.seek(0)

  ## Read the content
  print(file.read())
 
```
